Reshaping

```python
import numpy as np
```

```python
arr=np.array([[1,2,3],[4,5,6]])
newarr=arr.reshape(-1)  # converts to the 1-d array
print(newarr)
```

iterating through array


```python
arr=np.array([[1,2,3],[4,5,6]])
for i in arr:
    print(i)
```

```python
arr=np.array([1,2,3,4,5,6])
for i in np.nditer(arr,flags=['buffered'],op_dtypes=['S']):
    print(i)
```

```python
arr=np.array([[1,2,3],[4,5,6]])
for i in np.nditer(arr[:,::2]): #take all row & column with printing by skiping the index by 2
    print(i)
```

array concantenation


```python
arr1=np.array([[1,2,3],[4,5,6]])
arr2=np.array([[7,8,9],[10,11,12]])
final_arr=np.concatenate((arr1,arr2)) #its converts to multi-dimensional array
print("without Axis is :",final_arr)
final_arr2=np.concatenate((arr1,arr2),axis=0)
print("Axis with 0 is :",final_arr2)
final_arr3=np.concatenate((arr1,arr2),axis=1)
print("Axis with 1 is :",final_arr3)
```

```python
arr1=np.array([1,2,3,4,5,6])
arr2=np.array([7,8,9,10,11,12])
stacking=np.stack((arr1,arr2),axis=1)
print("stacking is:",stacking)
```
