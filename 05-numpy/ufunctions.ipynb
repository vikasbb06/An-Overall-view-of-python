Universal functions

```python
import numpy as np
```

```python
def add(a,b):
    return a+b
```

```python
#Here 3 parameters 1.function 2.nin(numpy input) 3.nout(numpy output)
yadd=np.frompyfunc(add,2,2)
print("The 2 array is:\n")
user_input=input("\nThe first array is:")
arr1=np.array(user_input.split(),dtype=int)
user_input2=input("\nThe second array is:")
arr2=np.array(user_input2.split(),dtype=int)
print("\nThe addition of 2 arrays from the function is:",add(arr1,arr2))
print("\nThe type of function is:",type(yadd))
```

```python
a=np.trunc([-3.1569,3.6564])
print("\nThe truncation is:",a)
b=np.ceil([-4.1919,5.0505])
print("The ceil function type is:",b)
arr=np.arange(1,10)
print("the value of logarithmic function is:",np.log10(arr)) # here we can give desired log to the base
arr1=np.array([2,4,6,8,10])
arr2=np.array([1,3,5,7,9])
c=np.sum([arr1,arr2],axis=1)
print("The sum function is:",c)
print("The product function is:",np.dot(arr1,arr2))
```
