using numpy and seaborn

```python
from random import random
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
sns.displot([1,2,3,4,5,6])
plt.show()
```

```python
sns.displot(np.random.normal(size=(1000)),kind="kde")
plt.show()
a=np.random.normal(loc=1,scale=2,size=(3,5)) # loc is the where the peak exists in gaussian distribution & scale is how flat the standard ditribution should be(standard deviation)
print("The array of desired parametres is:\n",a)
print("The graph of a above array is:\n")
sns.displot(a)
plt.show()
```

Binomial Distribution

```python
#binomial ditibution natins 3 parametres 1.number of elements and 2.probability 3.is the size of it
a=np.random.binomial(n=10,p=0.5,size=10)
print("The binomial distribution of size 10 is:\n",a)
sns.displot(a,kind="kde")
plt.show()
b=np.random.binomial(n=10,p=0.5,size=(4,5))
print("The binomial distribution of size (4,5) array is:\n",b)
sns.displot(b,kind="kde")
plt.show()
```

Normal & Binomial distribution together & finding the differene

```python
data={
    "Normal":np.random.normal(loc=2,scale=2,size=10),
    "Binomial":np.random.binomial(n=10,p=0.5,size=10)
}
print("The distribution of Normal and Binomial with size 10 is:\n",data.values())
sns.displot(data,kind="kde")
plt.show()
print("\nThe Difference is:If your number of trials (n) is large, a binomial distribution starts to look like a bell-shaped normal distribution.")
```

Poisson Distribution

```python
# it has 2 parameters 1.lambda=occurence of known event(constant=m) 2.size
a=np.random.poisson(lam=3,size=1000)
print("The poisson distribution of size 1000 is:\n",a)
sns.displot(a)
plt.show()
print("the graph in kde form is:\n")
sns.displot(a,kind="kde")
plt.show()
```

Normal & poisson distribution together & finding difference

```python
data={
    "Normal":np.random.normal(loc=40,scale=7,size=100),
    "Poisson":np.random.poisson(lam=40,size=100)
}
print("The distribution of Normal and Poisson with size 1000 is:\n",data.values())
sns.displot(data)
plt.show()
print("The graph in kde form is:\n")
sns.displot(data,kind="kde")
plt.show()
print("\nthe difference is:a Poisson distribution transforms into a normal distribution when its average rate(m) becomes large (typically m>20)")
```

binomial & poisson distribution together & finding difference

```python
data={
    "binomial":np.random.binomial(n=10,p=0.5,size=1000),
    "poison":np.random.poisson(lam=10,size=1000)
}
sns.displot(data,kind="kde")
plt.show()
print("\n The main difference is :A binomial distribution counts successes in a fixed number of independent trials, whereas a Poisson distribution counts the number of times a random event occurs")
```

Uniform Distribution. that has equal chance of getting random numbers in the given range

```python
#uniform takes 3 argument 1.low(lowest numbers) 2.high(high numbers that can come up) 3. size
sns.displot(np.random.uniform(size=(1000)),kind="kde")
plt.show()
print("\nThis shows the mult-dimension graph that is (3,5)")
sns.displot(np.random.uniform(size=(3,5)),kind="kde")
plt.show()
```

logistic distribution

```python
#it takes 3 parameters 1.loc 2.scale 3.size
a=np.random.logistic(loc=1,scale=2,size=(3,5))
print("The logistic distribution of size (3,5) is:\n",a)
sns.displot(a,kind="kde")
plt.show()
b=np.random.logistic(loc=1,scale=2,size=10)
print("The logistic distribution of size 10 is:\n",b)
sns.displot(b,kind="kde")
plt.show()
```

Normal & logistic distribution together & finding difference

```python
data={
    "Normal":np.random.normal(loc=2,scale=2,size=1000),
    "Logistic":np.random.logistic(loc=1,scale=2,size=1000)
}
sns.displot(data,kind="kde")
plt.show()
print("\nThe main difference:The normal distribution features a symmetric bell-shaped curve with thin exponential tails, whereas a logistic distribution has a similar bell shape but features heavier, fatter tails")
```
