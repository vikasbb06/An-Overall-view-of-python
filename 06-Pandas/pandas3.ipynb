```python
import pandas as pd
```

```python
df = pd.read_csv('results.csv')
df.head()
```

```python
# 2. Convert the 'Age' column from string to numeric, forcing errors to NaN if there is text like 'unknown'
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
# 3. Create the filter matching your print statement (> 10)
filt = (df['Age'] > 50)
# 4. Print the results
print("\nThe people of Age > 50 is:\n", df[filt])
```

```python
df["ConvertedCompYearly"]=pd.to_numeric(df["ConvertedCompYearly"], errors='coerce')
high_salary=(df["ConvertedCompYearly"]>7000)
df.loc[high_salary]
```

```python
print("\nThe continuation is:\n")
df.loc[high_salary,['Country','ConvertedCompYearly']]
```

```python
print("\nThe again updation is:\n")
countries=['India','United States of America','France']
#for comparing 'int' we can use '==' operator but for string we should use isin(parameter) for the usage
country_filt=df['Country'].isin(countries)
df.loc[country_filt,'Country']
```
