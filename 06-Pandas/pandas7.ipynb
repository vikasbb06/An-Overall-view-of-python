```python
import pandas as pd
na_vals=['NA','Missing']
df=pd.read_csv('results.csv',na_values=na_vals)
df.head(10)
```

```python
df['YearsCode'].unique()
```

```python
print(df['YearsCode'].dtypes)
df['YearsCode'] = pd.to_numeric(df['YearsCode'], errors='coerce')
print("Mean",df['YearsCode'].mean())
print("Median",df['YearsCode'].median())
print("Standard Deviation",df['YearsCode'].std())
```

```python
df.dropna(how='all', inplace=True)
df.dropna(subset=['SurveyLength'], inplace=True)
df.head(5)
```
