```python
import numpy as np
import pandas as pd

df=pd.read_csv('results.csv')
df.head()
```

```python
df_median=df['ConvertedCompYearly'].median()
print(df_median)
df_full=df.median(numeric_only=True)
print(df_full)
```

```python
#It describes the whole data;ad the quick over-view of the data
df.describe()
```

```python
#count method gives how many people answered the particular question
df_count=df['ConvertedCompYearly'].count()
print(df_count)
#value count gives the particular insights
df_value=df['SurveyEase'].value_counts()
print(df_value)
```

```python
country_grp=df.groupby('Country')
print(country_grp) #it returns the object created and address
print('United States of America' in country_grp.groups)
country_grp.get_group('United States of America')
```

```python
# Get data for the United States
us_data = country_grp.get_group('United States of America')
# Count the occurrences of each value in the Check column
us_data['Check'].value_counts()
```

```python
#To display the top country results
country_grp[['RemoteWork']].value_counts().head(5)
```

```python
#remote-work categories for each country
country_grp[['RemoteWork']].value_counts().groupby(level=0).head(5)
```

```python
#for the one particular country
ind_data=country_grp[['RemoteWork']].value_counts().loc['India']
print(ind_data)
filt=df["Country"]=="India"
data=df.loc[filt,"RemoteWork"].value_counts()
print(data)
print("this is the two methods of approching teh answer")
print("Median:",country_grp["ConvertedCompYearly"].median().head(5))
```

```python
#aggregator functions
agg_fun=country_grp["ConvertedCompYearly"].agg(['mean','median','sum'])
print(agg_fun.head(5))
co_agg=country_grp["ConvertedCompYearly"].agg(['mean','median']).loc['India']
print("The mean & median of the India is:\n",co_agg)
```
