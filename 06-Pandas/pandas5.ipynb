```python
import pandas as pd
```

```python
df=pd.read_csv('results.csv')
df.head()
```

Adding the new row

```python
df.loc[len(df)] = {
    'ResponseId': 100,
    'MainBranch': 'I am learning to code',
    'Age': '18-24 years old',
    'Employment': 'Student, full-time',
    'RemoteWork': 'Remote',
    'Check': 'Apples',
    'CodingActivities': 'Hobby',
    'EdLevel': 'Bachelor’s degree',
    'LearnCode': 'Books / Physical media'
}
```

Row operations 

```python
df = df.drop(index=0)
df = df.drop(index=[1, 2, 3])
df = df.drop(df[df['Age'] == 'Under 18 years old'].index)
df=df.reset_index(drop=True)
df.head()
```

Add a new column with the same value,using an existing column,at a specific position

```python
df['Country'] = 'India'
df['ResponseId_Double'] = df['ResponseId'] * 2
df.head()
```

Remove one column,multiple columns,a column using del

```python
df = df.drop(columns=['NewColumn'])
df = df.drop(columns=['Country', 'ResponseId_Double'])
del df['Check']
df.head()
```

Sorting

```python
# Sort by ResponseId in ascending order
df = df.sort_values(by='ResponseId')
# Sort by ResponseId in descending order
df = df.sort_values(by='ResponseId', ascending=False)
# Sort by Age alphabetically
df = df.sort_values(by='Age')
# Sort by multiple columns
df = df.sort_values(
    by=['Age', 'Employment'],
    ascending=[True, False]
)
# Reset index after sorting
df = df.reset_index(drop=True)
df.head()
```
