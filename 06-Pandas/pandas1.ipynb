```python
from asyncio.windows_events import NULL

import pandas as pd
import numpy as np
```

```python
my_dataset={
    "Cars":["Toyota","Audi","BMW","BENTLY"],
    'Passing':[3,6,7,4]
}
# conversion of messy data into a tabular form & dataframe is a  2-dimensional,mutuable data-structure
print(pd.DataFrame(my_dataset))
```

```python
data={
    "Days":["Monday","Tuesday","Wednesday","Thursday"],
    "CALORIES":[520,460,250,175],
    "DURATION":[50,45,30,30],
}
dt=pd.DataFrame(data)
print("The tabular form representation is:\n",dt)
print("The 1st row is :\n",dt.loc[0])
n=int(input("\nThe desired row elements data is:"))
print("The data is:\n",dt.loc[n])
```

```python
print("The 2-3 rows of desired is:\n",dt.loc[[2,3]])
df=pd.DataFrame(data,index=[1,2,3,4])
print("the tabular column form with index is:\n",df)
```

pandas Series:They are 1-D labeled array in python's pandas library

```python
a=np.array([11,15,19,21,25,29])
b=pd.Series(a)
print("the 1-D numpy array representation is:\n",b)
print("\nThe elements is accessed:",b[1])
c=pd.Series(a,index=['Mon','Tue','Wed','Thu','Fri','Sat'])
print("The tabular representation with changed row name is:\n",c)
data_set={
    'day1':78,'day2':56,'day3':69,'day4':75,'day5':99,
}
print("The Table is:\n",pd.Series(data_set))
print("\nThe key observation is it changes the key as index value.")
```

CSV file working

```python
datas=pd.read_csv('mark.csv')
print("\nThe info about data is:\n",datas.info())
print("The csv file read is:\n",datas)
print("\nthe No Rows is:",pd.options.display.max_rows)
#the to string give the entire data
print("The data from the bottom is:\n",datas.tail(10)) #last 10 is read
print("The deleting of the original dataset that contains 0 is:\n",datas.dropna(inplace=True))
print("The changing of the original dataset that contains 0  is:\n",datas.fillna(inplace=True,value=1))
print("The no of duplicates is:\n",datas.duplicated().sum())
print("The no of duplicates of same marks is:\n",datas.duplicated(subset='Mark').sum())
```
